```python
# Code generated by Claude.ai, access on October 25, 2023
# Prompt chain: generate python script that would be useful for data analysis; generate artificial data instead of loading a data file
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler  
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

np.random.seed(42)

n_samples = 1000
n_features = 20

X = np.random.rand(n_samples, n_features)
coefs = np.random.randn(n_features)
y = np.dot(X, coefs) > 0
y = y.astype(int)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

scaler = StandardScaler()  
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

model = LogisticRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print('Accuracy:', accuracy)
```

```python
# Put your AI-generated commented version in this cell.
# Make sure you cite which tool you used and the prompts you used to generate the comments.
# Tool used: ChatGPT 5.2, 07.25.2026
# Prompt used: "Give a version of the code that includes comments to explain what this code does."

import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Set random seed so the results are reproducible
np.random.seed(42)

# Set the number of samples and features
n_samples = 1000
n_features = 20

# Create random data
X = np.random.rand(n_samples, n_features)

# Create random coefficients
coefs = np.random.randn(n_features)

# Create class labels using the dot product
y = np.dot(X, coefs) > 0

# Convert True/False labels to 1/0 labels
y = y.astype(int)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Scale the features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Create and train the logistic regression model
model = LogisticRegression()
model.fit(X_train, y_train)

# Make predictions on the test data
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

# Print the result
print("Accuracy:", accuracy)
```

```python
# Put modified version #1 here
# Tool used: ChatGPT 5.2, 07.25.2026
# Prompt used: "Modify this code so the accuracy becomes lower by adding noise to the labels."

import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

np.random.seed(42)

n_samples = 1000
n_features = 20

X = np.random.rand(n_samples, n_features)
coefs = np.random.randn(n_features)

y = np.dot(X, coefs) > 0
y = y.astype(int)

# Flip some labels to add noise
noise_index = np.random.choice(n_samples, size=300, replace=False)
y[noise_index] = 1 - y[noise_index]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

model = LogisticRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy with noisy labels:", accuracy)
```

```python
# Put modified version #2 here
# Tool used: ChatGPT 5.2, 07.25.2026
# Prompt used: "Modify this code so it uses fewer features."

import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

np.random.seed(42)

n_samples = 1000
n_features = 5

X = np.random.rand(n_samples, n_features)
coefs = np.random.randn(n_features)

y = np.dot(X, coefs) > 0
y = y.astype(int)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

model = LogisticRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy with fewer features:", accuracy)
```

```python
# Put modified version #3 here
# Tool used: ChatGPT 5.2, 07.25.2026
# Prompt used: "Modify this code to use a decision tree classifier instead of logistic regression."

import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

np.random.seed(42)

n_samples = 1000
n_features = 20

X = np.random.rand(n_samples, n_features)
coefs = np.random.randn(n_features)

y = np.dot(X, coefs) > 0
y = y.astype(int)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Decision tree accuracy:", accuracy)
```
