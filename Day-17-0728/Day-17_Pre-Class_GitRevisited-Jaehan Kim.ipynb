### 1.2 Creating your first git "branch" 

Branches are a powerful and important tool when using git.  However, some folks are never exposed to this extra functionalty, thus missing out on their utility and power! Once you understand how they work you will likely find ways to use them all the time (sort of like git itself). 

Branches are a great location for doing new work in your git repositories. Branches are a safe place to store changes to files without fear of running into issues with work that others might be doing on the same repository.  Anytime you want to change files, it is good to get into the habit of making sure you are making and committing those changes to a branch.  

To create a branch just type the following:

```bash
    git branch BRANCHNAME
```

Once you have a new branch, you can list all of your branches (including the default "main" branch) and see which one you are on using the following command:

```bash
    git branch
    # Notice the lack of a BRANCHNAME here!
```

To switch between branches just use the ```git checkout``` command with the BRANCHNAME:

```bash
    git checkout BRANCHNAME
```

### 2.2 Git Clone vs Branch vs Fork

Here’s a short, conceptual summary of some different Git concepts that are important to understand when collaborating with others:

- **clone**: Make a full local copy of an existing remote repo (including history) on your machine so you can work on it directly. [stackoverflow](https://stackoverflow.com/questions/3329943/what-are-the-differences-between-git-branch-fork-fetch-merge-rebase-and-clon)
- **branch**: Create a parallel line of development within the *same* repo (local or remote) to isolate changes, then usually merge back. [graphite](https://graphite.com/guides/git-fork-vs-branch)
- **fork**: Create your *own* remote copy of someone else’s repo (e.g., on GitHub) under your account, then typically clone that to work locally and propose changes back via pull requests. [theserverside](https://www.theserverside.com/answer/Git-fork-vs-clone-Whats-the-difference)

Another way to think about these concepts is:

- **clone**: “Download the project to my computer.” You copy the whole repo (all files and history) from a server like GitHub to your local machine so you can work on it. 
- **branch**: “Create a sandbox inside this project.” You make a separate line of development *inside the same repo* so you can change code without breaking the main branch, then later merge it back. 
- **fork**: “Make my own copy on GitHub.” You create your own remote copy of someone else’s repo under your account, safe to edit as you like, then usually clone *that* and send changes back via pull requests.
